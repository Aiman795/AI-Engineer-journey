```python
class_info = pd.DataFrame({
    "pclass": [1, 2, 3],
    "class_name": ["First", "Second", "Third"],
})

port_info = pd.DataFrame({
    "embarked": ["C", "Q", "S"],
    "port_name": ["Cherbourg", "Queenstown", "Southampton"],
})
```

```python
pd.merge(class_info,df,on="pclass",how="left")
```

```python
pd.merge(port_info,df,on="embarked",how="right")
```

```python
len(pd.merge(port_info, df, on="embarked", how="right"))   # 1309
```

```python
len(pd.merge(port_info, df, on="embarked", how="left"))
```

```python
len(pd.merge(port_info, df, on="embarked", how="inner"))
```
